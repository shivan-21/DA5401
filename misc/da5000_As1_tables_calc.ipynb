```python
from math import sqrt, exp
import numpy as np
from scipy import stats, integrate

np.set_printoptions(precision=6, suppress=True)
```

### Question 1

```python
print("==== Question 1: Two-sample test (Welch) ====")
nX, nY = 30, 25
meanX, meanY = 8150.0, 7600.0
sX, sY = 900.0, 1100.0

se = sqrt((sX**2)/nX + (sY**2)/nY)
t_obs = (meanX - meanY) / se
nu = ((sX**2 / nX + sY**2 / nY)**2) / (
    ((sX**2 / nX)**2) / (nX - 1) + ((sY**2 / nY)**2) / (nY - 1)
)
t_crit_2sided = stats.t.ppf(0.975, df=nu)
p_val_two_sided = 2 * (1 - stats.t.cdf(abs(t_obs), df=nu))

print(f"t_obs = {t_obs:.6f}")
print(f"Welch df (nu) = {nu:.6f}")
print(f"Two-sided t_critical (alpha=0.05) = {t_crit_2sided:.6f}")
print(f"Two-sided p-value = {p_val_two_sided:.6f}")
```

### Question 2

```python
print("\n==== Question 2: One-sample t-test ====")
n = 25
xbar = 17.0
s = 6.0
mu0 = 15.0

t_one = (xbar - mu0) / (s / sqrt(n))
df_one = n - 1
t_crit_one = stats.t.ppf(0.95, df=df_one)  # one-sided, right-tail alpha=0.05
p_val_one_sided = 1 - stats.t.cdf(t_one, df=df_one)

print(f"t_obs = {t_one:.6f}, df = {df_one}")
print(f"One-sided critical t (alpha=0.05) = {t_crit_one:.6f}")
print(f"One-sided p-value = {p_val_one_sided:.6f}")
```

```python
print("\n==== Question 3: ±1 signal with Gaussian noise (error probability) ====")
def error_prob_gaussian(sigma):
    return stats.norm.cdf(-1.0 / sigma)  # Phi(-1/sigma)

for sigma in [0.5, 1.0, 2.0]:
    print(f"sigma={sigma}: P(error) = {error_prob_gaussian(sigma):.6f}")
print("General formula: P(error) = Phi(-1/sigma) = Q(1/sigma)")

```

```python
print("\n==== Question 4: Binomial test (n=20, p0=0.75), reject if X <= 10 ====")
n4, p0 = 20, 0.75
k_reject = 10
alpha = stats.binom.cdf(k_reject, n4, p0)
print(f"Significance level alpha = P(X <= {k_reject} | p={p0}) = {alpha:.6f}")

p_alt = 0.6
power_at_p_alt = stats.binom.cdf(k_reject, n4, p_alt)
print(f"Power at p = {p_alt} = P(X <= {k_reject} | p={p_alt}) = {power_at_p_alt:.6f}")
```

```python

print("\n==== Question 5: Joint PDF problem ====")
# f(x,y) = c x^2 y for 0 <= y <= x <= 1
c = 10.0
print(f"Constant c (analytic) = {c}")

# Marginals analytic:
# f_X(x) = 5 x^4 for 0<=x<=1
# f_Y(y) = (10/3) y (1 - y^3) for 0<=y<=1

def fX(x):
    return 5 * x**4 if 0 <= x <= 1 else 0.0

def fY(y):
    return (10.0/3.0) * y * (1 - y**3) if 0 <= y <= 1 else 0.0

# numeric checks by integration
print(f"Integral of f_X over [0,1] = {integrate.quad(lambda t: fX(t), 0, 1)[0]:.6f}")
print(f"Integral of f_Y over [0,1] = {integrate.quad(lambda t: fY(t), 0, 1)[0]:.6f}")

# (d) P(Y <= X/2)
integrand = lambda y, x: c * x**2 * y
p_y_le_x_over_2 = integrate.dblquad(integrand, 0, 1, lambda x: 0, lambda x: x/2)[0]
print(f"P(Y <= X/2) numeric = {p_y_le_x_over_2:.6f}  (analytic = 0.25)")

# (e) P(Y <= X/4 | Y <= X/2)
p_y_le_x_over_4 = integrate.dblquad(integrand, 0, 1, lambda x: 0, lambda x: x/4)[0]
cond_prob = p_y_le_x_over_4 / p_y_le_x_over_2
print(f"P(Y <= X/4) numeric = {p_y_le_x_over_4:.6f}  (analytic = 1/16 = 0.0625)")
print(f"Conditional P(Y <= X/4 | Y <= X/2) = {cond_prob:.6f}  (analytic = 0.25)")
```

```python

print("\n==== Question 6: Bessel's correction (numeric simulation check) ====")
np.random.seed(0)
mu = 2.0
sigma = 3.0
n_sim = 200000
n_sample = 10

data = np.random.normal(loc=mu, scale=sigma, size=(n_sim, n_sample))
sample_vars_unbiased = data.var(axis=1, ddof=1)  # denominator n-1
mean_of_sample_vars = sample_vars_unbiased.mean()
sample_vars_biased = data.var(axis=1, ddof=0)  # denominator n
mean_of_sample_vars_biased = sample_vars_biased.mean()

print(f"True variance sigma^2 = {sigma**2:.6f}")
print(f"Mean of unbiased sample variances (ddof=1) over {n_sim} sims = {mean_of_sample_vars:.6f}")
print(f"Mean of biased sample variances (ddof=0) over {n_sim} sims = {mean_of_sample_vars_biased:.6f}")
```

```python

print("\n==== Question 7: Poisson process (meteorites) ====")
lambda_per_day = 1.0 / 10.0
expected_wait = 1.0 / lambda_per_day
prob_at_least_one_in_12h = 1.0 - np.exp(-lambda_per_day * 0.5)
print(f"Rate lambda (per day) = {lambda_per_day:.6f}")
print(f"Expected waiting time = {expected_wait:.6f} days")
print(f"P(at least one meteorite in 12 hours) = {prob_at_least_one_in_12h:.6f}")
```
